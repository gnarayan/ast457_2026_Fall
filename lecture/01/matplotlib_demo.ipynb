## Real data: the same tools on actual catalogs

Everything above used made-up arrays. The same handful of commands is how real
figures get made. Two datasets live next to this notebook: `L14_halos.dat`
(dark-matter halos from an N-body simulation: x, y, z, mass) and
`L14_atnf_pulsars_20200112.dat` (the ATNF pulsar catalog, 2800 pulsars).
These examples are adapted from ASTR 310.

**Scatter with meaning**: point size carries mass, color carries a third
coordinate. This is the random-scatter demo from earlier, except now every
channel encodes something.

```python
x, y, z, m = np.loadtxt('L14_halos.dat', unpack=True, skiprows=3)

plt.figure(figsize=(6, 5))
sizes = 70*(np.log10(m) - 11.3)   # size ~ log mass (offset below the minimum, or sizes go negative)
plt.scatter(x, y, s=sizes, c=z, cmap='Spectral')
cbar = plt.colorbar()
cbar.set_label(r'$z$ (Mpc)')
plt.title('Projected halo positions')
plt.xlabel(r'$x$ (Mpc)')
plt.ylabel(r'$y$ (Mpc)')
plt.xlim(0., 89.8)
plt.ylim(0., 89.8)
```

**Log axes, selections, and overplotting**: the $P$–$\dot{P}$ diagram, the
H–R diagram of pulsar astronomy. Real catalogs have missing values.
`astropy.io.ascii` reads this fixed-width table into *masked* columns, and the
mask on `companion_type` is how we separate binaries from isolated pulsars.
Note the two populations: ordinary pulsars (upper right) and the
millisecond pulsars (lower left), which were spun back up by accretion from a
companion, which is why so many of them are orange.

```python
from astropy.io import ascii

data = ascii.read('L14_atnf_pulsars_20200112.dat', format='fixed_width', header_start=2)
P, Pdot, ctype = data['period'], data['period_dot'], data['companion_type']
binary = ~ctype.mask
notbin = ctype.mask

plt.figure(figsize=(6, 5))
plt.loglog(P[notbin], Pdot[notbin], 'k.', ms=4, label='Isolated')
plt.loglog(P[binary], Pdot[binary], 'C1o', ms=4, mfc='none', label='Binary')
ldeath = np.array([0.03, 40.])
plt.plot(ldeath, 7.14e-18*ldeath**3.2, 'k:', label='Death line')
plt.xlabel(r'$P\ ({\rm s})$')
plt.ylabel(r'$\dot{P}$')
plt.title('ATNF Pulsar Catalog')
plt.legend(loc='best')
plt.xlim([0.001, 40.])
plt.ylim([1.e-22, 1.e-8])
```

**Going 3-D**: the same catalog in space. `SkyCoord` converts the catalog's
RA/Dec/distance into Galactocentric x, y, z (more on coordinates in
`help/astro_conventions.md`), and a wireframe marks the Galactic disk. Most
pulsars hug the disk; the tight clumps well off the plane are globular
clusters. Drag to rotate (that is what `%matplotlib ipympl` buys you).

```python
from astropy.coordinates import SkyCoord

pos = SkyCoord(ra=data['ra'], dec=data['dec'], unit=('hour', 'deg', 'kpc'),
               distance=data['distance'])
pos = pos.transform_to('galactocentric')

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
ax.scatter(pos[notbin].x, pos[notbin].y, pos[notbin].z, c='k', s=5)
ax.scatter(pos[binary].x, pos[binary].y, pos[binary].z, c='C1', s=5)
r = np.linspace(0., 25., 13).reshape(-1, 1)   # kpc
phi = np.linspace(0., 2*np.pi, 181)
ax.plot_wireframe(r*np.cos(phi), r*np.sin(phi), np.zeros((13, 181)),
                  color='C0', linewidth=0.5)
ax.set_xlabel(r'$x$ (kpc)')
ax.set_ylabel(r'$y$ (kpc)')
ax.set_zlabel(r'$z$ (kpc)')
ax.set_title('ATNF pulsars, galactocentric')
fig.show()
```

**Saving figures**: `savefig` writes whatever format the extension implies.
Use a vector format (PDF, SVG) for papers and slides, and set `dpi` for
rasters; `bbox_inches='tight'` trims the whitespace padding.

```python
fig.savefig('pulsars_3d.png', dpi=200, bbox_inches='tight')
fig.savefig('pulsars_3d.pdf', bbox_inches='tight')   # vector, resolution-independent
```
