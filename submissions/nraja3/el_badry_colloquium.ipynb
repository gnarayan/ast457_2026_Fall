# Colloquium Notes
## Finding Black Holes and Neutron Stars with Gaia

- any massive O-type stars are predicted to leave behind black holes because of their short lifespans.
- Only a small percentage of the Milky Way's estimated **10^8 stellar-mass black holes** have been found.
- The majority of conventional findings originate from **X-ray binaries** where X-rays are produced when stuff from a normal star collides with a black hole. 
- Merging black-hole binaries are detected by gravitational-wave detectors.
- Question **where are all of the quiet black holes that are not actively accreting?**

### Gaia and astrometric binaries
- **Gaia** frequently made accurate measurements of the locations of stars throughout the sky.
- An astrometric orbit is created when a visible star circles an invisible object.
- Kepler's equations are used to calculate the mass of the unseen companion based on the orbit, parallax/distance, and mass of the visible star.
- Combining astrometry with radial velocities gives a better 3-D orbit and companion mass because ground-based **radial velocity** measurements include motion along our line of sight.Gaia DR3 generated around **170,000 binary orbits** compared to previous astrometric missions,.

### Gaia black holes
- Gaia DR3 led to two confirmed non-accreting black holes, often called **Gaia BH1 and BH2**.
- BH1, a visible star that resembles the Sun is in orbit around an invisible companion that is approximately **9.3 solar masses**.
- Because the huge black-hole progenitor should have grown into a red supergiant and maybe swallowed the low-mass partner, these systems are hard to explain using standard binary evolution.

### Masses, neutron stars, 
- Black holes and neutron stars have a relative shortage in the observed compact-object mass distribution which means many black holes are above **5 solar masses**, whereas the majority of neutron stars are below **2 solar masses**.
- Gaia also finds dark companions with masses around **1.3–2 solar masses**.
- Some of the metal-poor neutron-star companion stars show unusually strong **lithium enhancement**
- Gaia measurements also help study **natal kicks** from supernovae. 

### Why Gaia
- Gaia's longer time baseline allows longer-period and physically larger binary orbits to be measured.
- Because larger orbits can be detected at greater distances, the searchable volume increases dramatically.
- estimated that **DR4 could yield roughly 10 times more binaries** than DR3, which should greatly expand the sample of black holes, neutron stars, and other unseen companions.

### Questions
- What is the maximum distance Gaia can detect a non-accreting black hole around a Sun-like star?
- How strongly does metallicity influence the final mass of a black hole?
- How often do stellar encounters in clusters exchange black-hole companions?
- Why does Gaia have difficulty measuring binary systems with orbital periods close to one year?
